# calculate the confidence interval using T-procedure there are five point 
# 1. population STD = X 
# 2. population Mean = X 
# 3. apply (CLT) sample = 30 and times = 10 samples 
# 4. confidence level is 95% infrence 
# 5.formula = CI  T- procedure formula 

```python
df.shape
```

```python
df['Fare'].plot(kind='kde')
```

```python
sample =[]
stds = []
for i in range(10):
    x = df['Fare'].dropna().sample(30).values
    stds.append(x.std())
    sample.append(x.tolist())
```

```python
sample = np.array(sample) # convert the python list to np array 
```

```python
X_bar = sample.mean(axis=1) # accoring to formula calculate the all sample column wise 
```

```python
X_bar
```

```python
sample_std = np.mean(stds)  # calcualte the sample mean 
```

```python
# calulate the value 95% confidence interval level 
lower_limit = X_bar.mean() - 2.042 * (sample_std/np.sqrt(30))
upper_limit = X_bar.mean() + 2.042 * (sample_std/np.sqrt(30))
```
